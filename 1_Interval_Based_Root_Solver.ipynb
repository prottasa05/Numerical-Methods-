```python
# Initial Setup
xl = float(input("Enter Lower Limit: "))
xu = float(input("Enter Upper Limit: "))
n = int(input("Enter Significant Figure: "))
es = 0.5 * 10**(2-n)
Er = 100
it = 1
print("Es =", es, "%")
```

```python
def sf(num):
    return float(f"{num:.{n}g}")
```

```python
def f(x):
    return sf(x**3-4*x-9)
```

```python
while Er >= es:
    xr = sf((xl + xu) / 2)
    if it == 1:
        Ea = 0
        Er = 100
    else:
        Ea = sf(abs(xr - xrold))
        Er = sf(abs((xr - xrold) / xr) * 100)
    print("Iteration:", it)
    print("xl =", sf(xl))
    print("xu =", sf(xu))
    print("xr =", xr)
    print("Ea =", Ea)
    print("Er =", Er, "%")
    print()
    if f(xl) * f(xr) < 0:
        xu = xr
    else:
        xl = xr
    xrold = xr
    it = it + 1
```

```python
print("Final Root =", xr)
print("Number of Iterations =", it-1)
```
